# This script has all prepared cells for loading, preprocessing and saving processed data

```python
# libraries
import os
import numpy as np

# scripts
from helper_functions.A_loading_helpers import load_csv_data, save_numpy, load_numpy
import helper_functions.C_preprocessing_helpers as ph

# metadata
from data.feature_properties.cleaning_rules_continuous import cleaning_rules
```

## Loading Raw

```python
data_path = 'data'
x_train, x_test, y_train, train_ids, test_ids =  load_csv_data(data_path, sub_sample=False)
x_train_raw = x_train.copy()
x_train.shape
```

## Preprocessing

```python
# TRAINING DATA
X_proc_train, feature_names_proc, feature_dct_proc = ph.preprocessing_pipeline(
    x_train,
    feature_names_path="data/feature_properties/feature_names.csv",
    classes_path="data/feature_properties/feature_classes.json",
    cleaning_rules=cleaning_rules,
    drop_correlated_mode='missing'
)
```

```python
# TEST DATA
X_proc_test, feature_names_proc, feature_dct_proc = ph.preprocessing_pipeline(
    x_test,
    feature_names_path="data/feature_properties/feature_names.csv",
    classes_path="data/feature_properties/feature_classes.json",
    cleaning_rules=cleaning_rules,
    drop_correlated_mode='missing'
)
```

# Save everything as npy files

```python
procesed_data_path = 'data/processed/'
train_path = procesed_data_path + 'x_train_processed.npy'
test_path = procesed_data_path +'x_test_processed.npy'

save_numpy(X_proc_train, train_path)
save_numpy(X_proc_test, test_path)

# labels

y_train_path = procesed_data_path + 'y_train.npy'
save_numpy(y_train, y_train_path)
```
